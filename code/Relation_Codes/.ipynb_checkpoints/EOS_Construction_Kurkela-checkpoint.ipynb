```python
import numpy as np
import matplotlib.pyplot as plt
from numba import jit
from scipy.optimize import minimize
from scipy import interpolate
```

$
P_{i} = k_{i}(n_{i})^{\Gamma_{i}}
$

```python
@jit(nopython=True)
def P_from_nb(n_list, K, Gam):
    p = K*(n_list)**(Gam)
    return p

@jit(nopython=True)
def mu_from_nb(n_list, K, Gam, m0):
    t1 = ((K*Gam)/(Gam - 1))*((n_list)**(Gam - 1))
    re = t1 + m0
    return re

@jit(nopython=True)
def e_from_n(n_list, K, Gam, m0):
    t1 = (K * ((n_list) ** Gam)) / (Gam - 1)
    t2 = t1 + (m0 * (n_list))
    return t2

@jit(nopython=True)
def cs_from_n(mu,K,Gam,n_list):
    t1 = 1/mu
    t2 = K*Gam*((n_list)**(Gam-1))
    t = t1*t2
    return t

@jit(nopython=True)
def k(p_initial,n_initial,gam):
    result = p_initial/(n_initial**gam)
    return result

@jit(nopython=True)
def calculate_m0(energy_density, K, n, Gam):
    return (energy_density - (K * (n ** Gam)) / (Gam - 1)) / n
```

$
n_{0} = 0.16 fm^{-3} \\
n_{t} = [1.1,10] n_{0} \\
\Delta n = [0,9.09] n_{t}
$

```python
n0 = 0.16

def ntt(z):
    return np.linspace(1.1*0.16,10*0.16,z)

def delta_nt(nt,z):
    return np.linspace(0*nt,9.09*nt,z)
```

$
\Gamma_{h} = [1.1,4] \\
\Gamma_{Q} = [0.5,4]
$

```python
def gamm_h(z):
    return np.linspace(1.1,4,z)
def gamm_q(z):
    return np.linspace(1.1,3,z)
```

$
Seidov - Limit : \\
\frac{\Delta \epsilon}{\epsilon_{c}} \geq \frac{1}{2} + \frac{3}{2} \frac{P_{c}}{\epsilon_{c}}
$


```python
def seidov_limit_constraint(delta_epsilon, epsilon_c, P_c):
    lhs = delta_epsilon / epsilon_c
    rhs = 0.5 + 1.5 * (P_c / epsilon_c)
    
    # Return True if lhs is greater than or equal to rhs, else False
    return lhs >= rhs
```

```python
import itertools

def constructing_EOS(n, y):

    p, mu, e, cs, n_total = [], [], [], [], []
    p_c_total, mu_c_total = [], []
    gam_h, gam_q =[], []
    n_transition, delta_n = [], []
    z = n 
    
    #gamma_h = gamm_h(z)
    gamma_q = gamm_q(z)

    gamma_h = [4.921]  # Instead of gamm_h(z)
    gamma_q = [6.0] 
    #gamma_h = [2.50,2.75,3.0,3.25,3.50]
    nt = [0.32]
    #print(nt)
    # Generate all combinations of gamma_h, gamma_q, nt, delta_n using itertools.product
    for g_h, g_q, n_t in itertools.product(gamma_h, gamma_q, nt):
        delta_n_11=[]
        #print("n")

        # Generate delta_n for each combination of n_t
        #d_n_list = delta_nt(n_t, 4)
        d_n_list = [0.42]
        
        # Iterate over delta_n values
        for d_n in d_n_list:
            #print(d_n)
            pih = 3.542
            muih = 977.5
            eih = 1.1 * 0.16 * muih - pih
            k_had = pih / (1.1 * 0.16) ** g_h

            n_had = np.linspace(1.1 * 0.16, n_t, y)

            p_had = P_from_nb(n_had, k_had, Gam=g_h)
            m_had = calculate_m0(energy_density=eih, K=k_had, n=1.1*0.16, Gam=g_h)
            mu_had = mu_from_nb(n_list=n_had, K=k_had, Gam=g_h, m0=m_had)
            e_had = e_from_n(n_list=n_had, K=k_had, Gam=g_h, m0=m_had)
            cs_had = cs_from_n(mu=mu_had, K=k_had, Gam=g_h, n_list=n_had)

            if cs_had[-1] < 1 :

                n_mid = np.linspace(n_t, n_t + d_n, y)

                p_mid = np.linspace(p_had[-1],p_had[-1],y)
                mu_mid = np.linspace(mu_had[-1],mu_had[-1],y)
                e_mid = n_mid*mu_mid - p_mid
                cs_mid = np.gradient(p_mid,e_mid)

                delta_e = e_mid[-1] - e_mid[0]
                p_c = p_had[-1]
                mu_c = mu_had[-1]
                e_c = e_had[-1]
                #truth_value = seidov_limit_constraint(delta_epsilon=delta_e, epsilon_c=e_c, P_c=p_c)
                truth_value =True

                if truth_value == True:
                    # If the Seidov limit constraint is satisfied
                    n_iq = n_t + d_n
                    piq = p_had[-1]
                    k_quark = piq / (n_iq ** g_q)
                    m_quark = calculate_m0(energy_density=e_mid[-1], K=k_quark, n=n_iq, Gam=g_q)
                    #n_fq = (1000 / k_quark) ** (1 / g_q)
                    #n_fq = 10*n0
                    n_quark = np.linspace(n_t + d_n, 3, y)

                    p_quark = P_from_nb(n_quark, k_quark, Gam=g_q)
                    mu_quark = mu_from_nb(n_list=n_quark, K=k_quark, Gam=g_q, m0=m_quark)
                    e_quark = e_from_n(n_list=n_quark, K=k_quark, Gam=g_q, m0=m_quark)
                    cs_quark = cs_from_n(mu=mu_quark, K=k_quark, Gam=g_q, n_list=n_quark)

                    p.append(np.concatenate((p_had, p_quark)))
                    mu.append(np.concatenate((mu_had, mu_quark)))
                    e.append(np.concatenate((e_had, e_quark)))
                    cs.append(np.concatenate((cs_had, cs_quark)))
                    n_total.append(np.concatenate((n_had, n_quark)))

                    gam_h.append(g_h)
                    gam_q.append(g_q)

                    n_transition.append(nt)
                    delta_n_11.append(d_n)

                    p_c_total.append(p_c)
                    mu_c_total.append(mu_c)
                delta_n.append(delta_n_11)
    # Default return if no valid result found
    return p, mu, e, cs, n_total, mu_c_total, p_c_total, gam_h, gam_q, n_transition, delta_n
```

```python
from itertools import product
import numpy as np
def find_first_larger_index(lst, value):
    return next((i for i, x in enumerate(lst) if x > value), None)
@jit(nopython=True)
def quad_P(mu_c, mu, alpha, beta, delP, P_c):
    t1 = alpha*(mu - mu_c)**2
    t2 = beta*(mu - mu_c)
    t3 = (1+delP)*P_c
    re = t1 + t2 + t3
    return re

@jit(nopython=True)
def alpha(delP, P_q, P_h, P_c, mu_c, mu_c_h, mu_c_q):
    if (mu_c_q - mu_c_h) == 0 or (mu_c_h - mu_c) == 0 or (mu_c_q - mu_c) == 0:
        re = -1
    else:
        t1 = (P_q - (1+delP)*P_c)/(mu_c_q - mu_c)
        t2 = (P_h - (1+delP)*P_c)/(mu_c_h - mu_c)
        re = (t1 - t2)/(mu_c_q - mu_c_h)
    return re

@jit(nopython=True)
def beta(delP, P_q, P_h, P_c, mu_c, mu_c_h, mu_c_q):
    if (mu_c_q - mu_c_h) == 0 or (mu_c_h - mu_c) == 0 or (mu_c_q - mu_c) == 0:
        re = -1
    else:
        t1 = (P_q - (1+delP)*P_c)/(mu_c_q - mu_c)
        t2 = (P_h - (1+delP)*P_c)/(mu_c_h - mu_c)
        t3 = (P_q - P_h)/(mu_c_q - mu_c_h)
        re = t1 + t2 - t3
    return re

def find_valid_parameters(delta_P_arr, y_intersect, x_intersect, p_total, mu_total):
    A, B = [], []
    mu_c_h_arr, mu_c_q_arr = [], []
    P_c_h_arr, P_c_q_arr = [], []
    delta_P_found = []

    index = find_first_larger_index(p_total, y_intersect)
    #print(index)
    x1 = np.array(mu_total[:index])
    y1 = np.array(p_total[:index])

    x2 = np.array(mu_total[index:])
    y2 = np.array(p_total[index:])

    P_h_int_func = interpolate.interp1d(x1, y1, fill_value="extrapolate")
    P_q_int_func = interpolate.interp1d(x2, y2, fill_value="extrapolate")

    mu_start = x_intersect-100
    mu_end = x_intersect+100

    mu_new = np.linspace(mu_start, mu_end, 1000)
    P_had_array = P_h_int_func(mu_new)
    P_quark_array = P_q_int_func(mu_new)
    #plt.plot(x1,y1)
    #plt.plot(x2,y2)
    #plt.show()
    #plt.plot(mu_new,P_had_array)
    #plt.plot(mu_new,P_quark_array)
    #plt.show()
    y_diff = np.abs(P_had_array - P_quark_array)

    for delta_P in delta_P_arr:
        found = False

        # Iterate through all combinations of mu_c_h and mu_c_q
        for mu_c_h, mu_c_q in product(mu_new, repeat=2):
            if found:
                break

            P_h = P_h_int_func(mu_c_h)
            P_q = P_q_int_func(mu_c_q)

            alp = alpha(delta_P, P_q, P_h, y_intersect, x_intersect, mu_c_h, mu_c_q)
            bet = beta(delta_P, P_q, P_h, y_intersect, x_intersect, mu_c_h, mu_c_q)

            # Check if alpha is positive
            if alp <= 0:
                continue

            # Calculate the mixed phase pressure
            P_m = quad_P(x_intersect, mu_new, alp, bet, delta_P, y_intersect)

            # Split the pressure difference calculation
            min_idx = np.argmin(y_diff)
            m_h_P_diff = P_m[:min_idx] - P_had_array[:min_idx]
            m_q_P_diff = P_m[min_idx:] - P_quark_array[min_idx:]

            # Validate the pressure conditions
            if np.all(m_h_P_diff >= 0) and np.all(m_q_P_diff >= 0):
                #print(np.min(m_h_P_diff), np.min(m_q_P_diff))
                delta_P_found.append(delta_P)
                A.append(alp)
                B.append(bet)
                mu_c_h_arr.append(mu_c_h)
                mu_c_q_arr.append(mu_c_q)
                P_c_h_arr.append(P_h)
                P_c_q_arr.append(P_q)
                found = True

    print(f"Found {len(A)} valid sets of parameters.")
    return A, B, mu_c_h_arr, mu_c_q_arr, P_c_h_arr, P_c_q_arr, delta_P_found
```

```python
def gradient(n, m):
    
    if len(n) != len(m):
        raise ValueError("n and m must have the same length.")
    
    gradients = []
    for i in range(len(n) - 1):  # Loop up to the second-to-last element
        grad = (n[i + 1] - n[i]) / (m[i + 1] - m[i])
        gradients.append(grad)
    
    return gradients

def compute_n_e(P, mu):
    """Compute number density (n_quad) and energy density (e_quad) from pressure (P) and chemical potential (mu)."""
    n_quad = gradient(P, mu)
    e_quad = (n_quad * mu[:-1]) - P[:-1]
    cs_quad = gradient(P[:-1],e_quad)
    return n_quad, e_quad, cs_quad


from TOV_Solver import generate_mr_curve
from EOS_construction import constructing_EOS
```

```python
def mixed_phase(mu_total, p_total,n_total, A, B, mu_c_h_arr, mu_c_q_arr, P_c_h_arr, P_c_q_arr, mu_crit, P_crit):
    P_arr_dp = []
    E_arr_dp = []
    number_density_dp = []
    mu_arr_dp = []

    for i in range(len(A)):
        index_1 = find_first_larger_index(mu_total, mu_c_h_arr[i])
        index_2 = find_first_larger_index(mu_total, mu_c_q_arr[i])

        mu_mixed = np.linspace(mu_total[index_1], mu_total[index_2], len(mu_total[index_1:index_2]))

        P_m = quad_P(mu_crit, mu_mixed, A[i], B[i], delta_P_arr[i], P_crit)
        P_had_half = p_total[:index_1]
        P_quark_half = p_total[index_2:]

        grad_had = n_total[:index_1]
        grad_m = gradient(P_m,mu_mixed)
        grad_quark = n_total[index_2:]

        mu_arr = np.concatenate((mu_total[:index_1], mu_mixed[:-1], mu_total[index_2:]))

        grad_dummy = np.concatenate((grad_had,grad_m,grad_quark))
        P_arr_dummy = np.concatenate((P_had_half,P_m[:-1],P_quark_half))
        e_dummy = (grad_dummy*mu_arr)-P_arr_dummy

        P_arr_dp.append(P_arr_dummy)
        E_arr_dp.append(e_dummy)
        number_density_dp.append(grad_dummy)
        mu_arr_dp.append(mu_arr)

    return P_arr_dp, E_arr_dp, number_density_dp, mu_arr_dp
```

```python
# Initialize lists to store all optimal parameters and distances for each index
a_optimal_2 = []
b_optimal_2 = []
P_optimal_2 = []
min_distance_points_2 = []
min_distances_2 = []

P_all = []
P_all_1 = []
P_max_points = []
P_max_points_1 = []

# Iterate through each mu_total and p_total index
for index in range(len(mu_total)):

    P_crit_1 = P_crit[index] * (1 + 0.01)
    #P_crit_2 = p_total[index][-3]
    P_crit_2 = P_crit[index] * (1 + 1)
    P_crit_new = np.linspace(P_crit_1, P_crit_2, 1000)

    b_optimal = []
    a_optimal = []
    P_optimal = []
    min_distance_points = []
    min_distances = []

    for P in P_crit_new:
        # Define your points
        x1, y1 = mu_total[index][0], p_total[index][0]  # First point the curve must pass through
        x2, y2 = mu_crit[index], P   # Critical point

        # For inequality constraint defining hadronic part.
        x_values = np.array(mu_total[index][1:])  # Hadronic part x_values
        f_values = np.array(p_total[index][1:])   # Corresponding y_values

        # Define the quadratic function
        def quadratic(x, a, b):
            return a * (x - x2)**2 + b * (x - x2) + y2

        # Objective function: sum of squared differences (least distance point)
        def objective(params):
            a, b = params
            return 0

        # Initial guess for parameters
        initial_guess = [0, 0]

        # Constraint: quadratic must pass through the first point (x1, y1)
        def eq_constraint(params):
            a, b = params
            return quadratic(x1, a, b) - y1

        # Inequality constraint: ensure quadratic stays above hadronic part
        def ineq_constraint(params):
            a, b = params
            return quadratic(x_values, a, b) - f_values

        # Define the constraints
        constraints = [
            {'type': 'eq', 'fun': eq_constraint},  # Must pass through (x1, y1)
            {'type': 'ineq', 'fun': ineq_constraint},  # Must stay above hadronic data
        ]

        # Perform the optimization
        result = minimize(objective, initial_guess, constraints=constraints)

        if result.success:
            a_opt, b_opt = result.x

            # Find the least distance point
            min_distance = float('inf')
            closest_point = None

            # Calculate the quadratic value and find minimum distance
            for x, f in zip(x_values, f_values):
                y_quad = quadratic(x, a_opt, b_opt)
                distance = np.abs(y_quad - f)  # Only y-direction distance
                if distance < min_distance:
                    min_distance = distance
                    closest_point = (x, y_quad)

            # Save the optimal solution
            a_optimal.append(a_opt)
            b_optimal.append(b_opt)
            P_optimal.append(P)
            min_distance_points.append(closest_point)
            min_distances.append(min_distance)

        else:
            continue

    # Append the current index's results to the global lists
    a_optimal_2.append(a_optimal)
    b_optimal_2.append(b_optimal)
    P_optimal_2.append(P_optimal)
    min_distance_points_2.append(min_distance_points)
    min_distances_2.append(min_distances)

    # Convert to arrays for processing
    min_distances_arr = np.array(min_distances)
    P_optimal_arr = np.array(P_optimal)
    min_distance_points_arr = np.array(min_distance_points)

    # Check if arrays are not empty
    if min_distances_arr.size > 0 and P_optimal_arr.size > 0 and min_distance_points_arr.size > 0:
        # Find the index of the minimum distance
        index_min = min_distances_arr.argmin()
        P_all.append(P_optimal_arr[index_min])
        P_all_1.append(P_optimal_arr[0])
        P_max_points.append(min_distance_points_arr[index_min])
        P_max_points_1.append(min_distance_points_arr[0])
    else:
        # Append zeros if arrays are empty
        P_all.append(0)
        P_all_1.append(0)
        P_max_points.append(0)
        P_max_points_1.append(0)
```

```python
P_all = []
P_all_1 = []
P_max_points = []
P_max_points_1 = []

# Iterate through each mu_total and p_total index
for index in range(len(mu_total)):

    P_crit_1 = P_crit[index]*(1+0.01)
    P_crit_2 = p_total[index][-3]
    P_crit_new = np.linspace(P_crit_1,P_crit_2,1000)

    b_optimal = []
    a_optimal = []
    P_optimal = []
    min_distance_points = []
    min_distances = []

    for P in P_crit_new:
        # Define your points
        x1, y1 = mu_total[index][0], p_total[index][0]  # First point the curve must pass through
        x2, y2 = mu_crit[index],P   # Critical point

        # For inequality constraint defining hadronic part.
        x_values = np.array(mu_total[index][1:])  # Hadronic part x_values
        f_values = np.array(p_total[index][1:])   # Corresponding y_values

        # Define the quadratic function
        def quadratic(x, a, b):
            return a * (x - x2)**2 + b * (x - x2) + y2

        # Objective function: sum of squared differences (least distance point)
        def objective(params):
            a, b = params
            return 0

        # Initial guess for parameters
        initial_guess = [0, 0]

        # Constraint: quadratic must pass through the first point (x1, y1)
        def eq_constraint(params):
            a, b = params
            return quadratic(x1, a, b) - y1

        # Inequality constraint: ensure quadratic stays above hadronic part
        def ineq_constraint(params):
            a, b = params
            return quadratic(x_values, a, b) - f_values

        # Define the constraints
        constraints = [
            {'type': 'eq', 'fun': eq_constraint},  # Must pass through (x1, y1)
            {'type': 'ineq', 'fun': ineq_constraint},  # Must stay above hadronic data
        ]

        # Perform the optimization
        result = minimize(objective, initial_guess, constraints=constraints)

        if result.success:
            a_opt, b_opt = result.x

            # Find the least distance point
            min_distance = float('inf')
            closest_point = None

            # Calculate the quadratic value and find minimum distance
            for x, f in zip(x_values, f_values):
                y_quad = quadratic(x, a_opt, b_opt)
                distance = np.abs(y_quad - f)  # Only y-direction distance
                if distance < min_distance:
                    min_distance = distance
                    closest_point = (x, y_quad)

            # Save the optimal solution
            a_optimal.append(a_opt)
            b_optimal.append(b_opt)
            P_optimal.append(P)
            min_distance_points.append(closest_point)
            min_distances.append(min_distance)

        else:
            continue
    # Convert min_distances to a NumPy array
    min_distances = np.array(min_distances)
    P_optimal = np.array(P_optimal)
    min_distance_points = np.array(min_distance_points)


    # Check if min_distances and P_optimal contain values
    if min_distances.size > 0 and P_optimal.size > 0 and min_distance_points.size > 0:
        # Find the index of the minimum distance across all runs
        index = min_distances.argmin()

        # Save a specific optimal value to P_all based on index
        P_all.append(P_optimal[index])
        P_all_1.append(P_optimal[0])
        P_max_points.append(min_distance_points[index])
        P_max_points_1.append(min_distance_points[0])
    else:
        # Append zeros if arrays are empty
        P_all.append(0)
        P_all_1.append(0)
        P_max_points.append(0)
        P_max_points_1.append(0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define baryon number density
rho_B = np.linspace(0.25, 0.7, 500)

# Mock E/A functions based on behavior in the plot
EA_hadron = 940 + 70 * rho_B        # straight-ish
EA_quark = 970 + 30 * rho_B**1.2    # curves up
EA_MP = EA_hadron.copy()

# Add a "mixed phase" minimum around ρ ~ 0.45–0.55
EA_MP[(rho_B > 0.35) & (rho_B < 0.6)] -= 5 * np.exp(-((rho_B[(rho_B > 0.35) & (rho_B < 0.6)] - 0.47)**2) / 0.002)

# Pressure using thermodynamic relation: P = ρ^2 * d(E/A)/dρ
def pressure(rho, EA):
    dEA_drho = np.gradient(EA, rho)
    return rho**2 * dEA_drho

P_hadron = pressure(rho_B, EA_hadron)
P_quark = pressure(rho_B, EA_quark)
P_MP = pressure(rho_B, EA_MP)

# Plotting
plt.figure(figsize=(8, 5))
plt.plot(rho_B, P_hadron, label="Hadronic Phase", linestyle="--", color="gray")
plt.plot(rho_B, P_quark, label="Quark Phase", linestyle="--", color="brown")
plt.plot(rho_B, P_MP, label="Mixed Phase", color="blue", linewidth=2)

plt.xlabel("Baryon Number Density $\\rho_B$ [fm$^{-3}$]")
plt.ylabel("Pressure $P$ [MeV/fm$^3$]")
plt.title("Pressure vs. Baryon Number Density")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
